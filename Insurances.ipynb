```python
import pandas as pd
import numpy as np
```

```python
# 模拟生成保单数据
np.random.seed(42)
policy_data = {
    'policy_id': range(1, 101),
    'total_premium': np.random.randint(1000, 10000, 100),  # 总保费
    'first_year_premium_periodic': np.random.randint(0, 5000, 100),  # 新单首年期缴原/分保费收入 
    'first_year_premium_lump': np.random.randint(0, 5000, 100)  # 趸缴原/分保费收入
}
policy_df = pd.DataFrame(policy_data)

# 模拟生成再保分出业务明细数据
reinsurance_data = {
    'policy_id': range(1, 101),
    'ceded_premium': np.random.randint(0, 2000, 100)  # 分出保费
}
reinsurance_df = pd.DataFrame(reinsurance_data)

# 模拟生成机构数据
institution_data = {
    'institution_id': [1],
    'registered_capital': [500000]  # 注册资本
}
institution_df = pd.DataFrame(institution_data)

# 模拟保险市场原/分保费收入数据
market_premium_data = {
    'period': ['base', 'report'],
    'market_premium': [1000000, 1200000]
}
market_premium_df = pd.DataFrame(market_premium_data)
```

```python
merged_df = pd.merge(policy_df, reinsurance_df, on='policy_id')
```
